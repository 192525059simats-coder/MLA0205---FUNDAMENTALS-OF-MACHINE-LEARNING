```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Sample dataset
data = {
    'Height': [150, 160, 170, 180, 155, 165, 175, 185, 158, 178],
    'Weight': [50, 55, 65, 80, 52, 60, 75, 85, 54, 78],
    'Class': ['A', 'A', 'B', 'B', 'A', 'A', 'B', 'B', 'A', 'B']
}

df = pd.DataFrame(data)

# Separate input and output
X = df[['Height', 'Weight']]
y = df['Class']

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Create K-NN model
model = KNeighborsClassifier(n_neighbors=3)

# Train model
model.fit(X_train, y_train)

# Predict test data
y_pred = model.predict(X_test)

# Display results
print("Actual Values:", list(y_test))
print("Predicted Values:", list(y_pred))

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("\nAccuracy:", accuracy * 100, "%")

# Predict a new sample (with feature names)
new_sample = pd.DataFrame(
    [[172, 68]],
    columns=['Height', 'Weight']
)

result = model.predict(new_sample)

print("\nNew Sample:", new_sample.values.tolist())
print("Predicted Class:", result[0])
```
