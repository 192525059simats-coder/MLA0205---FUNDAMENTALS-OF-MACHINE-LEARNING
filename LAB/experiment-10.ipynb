```python
import numpy as np
from sklearn.mixture import GaussianMixture

# Sample dataset
X = np.array([
    [1], [2], [3], [4], [5],
    [10], [11], [12], [13], [14]
])

# Create EM model
model = GaussianMixture(n_components=2, random_state=42)

# Train model
model.fit(X)

# Predict clusters
labels = model.predict(X)

# Display results
print("Cluster Means:")
print(model.means_)

print("\nCluster Labels:")
print(labels)

print("\nModel Converged:")
print(model.converged_)
```
